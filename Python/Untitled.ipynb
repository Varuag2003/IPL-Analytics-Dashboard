## Step 7 - Remove Duplicate Records

```python
matches_clean.drop_duplicates(inplace=True)
ball_clean.drop_duplicates(inplace=True)
players_clean.drop_duplicates(inplace=True)
teams_clean.drop_duplicates(inplace=True)

print("Duplicate records removed successfully ✅")
```

## Step 8 - Analyze Missing Values

```python
# Count missing values

missing_values = matches_clean.isnull().sum()

# Show only columns having missing values
missing_values = missing_values[missing_values > 0]

missing_values
```

```python
# Percentage of missing values

missing_percentage = (matches_clean.isnull().sum() / len(matches_clean)) * 100

missing_percentage = missing_percentage[missing_percentage > 0]

missing_percentage.sort_values(ascending=False)
```

## Step 9 - Handle Missing Values

```python
# Fill missing values

matches_clean["city"] = matches_clean["city"].fillna("Unknown")

matches_clean["match_number"] = matches_clean["match_number"].fillna("League Match")

matches_clean["player_of_match"] = matches_clean["player_of_match"].fillna("Not Awarded")

print("Missing values handled successfully ✅")
```

```python
matches_clean.isnull().sum()
```

## Step 10 - Handle Match Result Columns

```python
matches_clean["win_by_runs"] = matches_clean["win_by_runs"].fillna(0)
matches_clean["win_by_wickets"] = matches_clean["win_by_wickets"].fillna(0)

print("Match result columns cleaned successfully ✅")
```

### Which phase of the innings produces the most runs?

```python
ball_clean["Phase"] = ball_clean["over"].apply(
    lambda x: "Powerplay"
    if x <= 6
    else (
        "Middle Overs"
        if x <= 15
        else "Death Overs"
    )
)

phase_runs = (
    ball_clean.groupby("Phase")["total_runs"]
    .sum()
    .reset_index()
)

phase_runs
```

```python
ball_clean["Phase"] = ball_clean["over_number"].apply(
    lambda x: "Powerplay"
    if x <= 6
    else (
        "Middle Overs"
        if x <= 15
        else "Death Overs"
    )
)

phase_runs = (
    ball_clean.groupby("Phase")["total_runs"]
    .sum()
    .reset_index()
)

phase_runs
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    phase_runs["Phase"],
    phase_runs["total_runs"]
)

plt.title("Runs by Match Phase")
plt.xlabel("Phase")
plt.ylabel("Runs")

plt.show()
```

### Which IPL teams have the highest win percentage?

```python
# Matches Played by each Team

matches_played = pd.concat(
    [matches_clean["team1"], matches_clean["team2"]]
).value_counts()

# Matches Won

matches_won = matches_clean["match_winner"].value_counts()

# Win Percentage

win_percentage = (
    (matches_won / matches_played) * 100
).round(2)

win_percentage = (
    win_percentage
    .sort_values(ascending=False)
    .reset_index()
)

win_percentage.columns = ["Team", "Win Percentage"]

win_percentage
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    win_percentage["Team"][:10],
    win_percentage["Win Percentage"][:10]
)

plt.title("Top 10 Teams by Win Percentage")
plt.xlabel("Team")
plt.ylabel("Win Percentage (%)")

plt.xticks(rotation=70)

plt.show()
```

### Insight

- Teams with higher win percentages have maintained better consistency over multiple IPL seasons.
- This metric compares team performance relative to the number of matches played.
